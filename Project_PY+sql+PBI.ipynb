```python
# Checking if missing data or null values are present in the dataset
df.isnull().sum()
```

```python
# Filling the null value of Review Rating column with the median rating of the product category
df['Review Rating']= df.groupby('Category')['Review Rating'].transform(lambda x: x.fillna(x.median()))
```

```python
df.isnull().sum()
```

```python
# Renaming columns according to snake casing for better readability and documentation

df.columns = df.columns.str.lower()
df.columns = df.columns.str.replace(' ','_')
df = df.rename(columns={'purchase_amount_(usd)':'purchase_amount'})
```

```python
df.columns
```

```python
# create a new column age_group 
labels = ['Young Adult', 'Adult', 'Middle-aged', 'Senior']
df['age_group'] = pd.qcut(df['age'], q=4, labels = labels)   # qcut splits the ages into 4 equal sized group based on data distribution
df[['age','age_group']].head(10)
```

```python
# create new column purchase_frequency_days

frequency_mapping = {
    'Fortnightly': 14,
    'Weekly': 7,
    'Monthly': 30,
    'Quarterly': 90,
    'Bi-Weekly': 14,
    'Annually': 365,
    'Every 3 Months': 90
}

df['purchase_frequency_days'] = df['frequency_of_purchases'].map(frequency_mapping)
```

```python
df[['purchase_frequency_days','frequency_of_purchases']].head(10)
```

```python
df[['discount_applied','promo_code_used']].head(10)
```

```python
# Dropping promo code used column
df = df.drop('promo_code_used', axis=1)
```

### Data Visualization using Matplotlib and Seaborn Libraries

```python
df.columns
```

```python
# Bar Chart – Total Purchase Amount by Category
import matplotlib.pyplot as plt

# Group data
category_sales = df.groupby("category")["purchase_amount"].sum()

# Plot
plt.figure(figsize=(6,4))
plt.bar(category_sales.index, category_sales.values)

plt.title("Total Purchase Amount by Category")
plt.xlabel("Category")
plt.ylabel("Purchase Amount")
plt.xticks(rotation=45)

plt.show()
```

```python
# Pie Chart – Gender Distribution
import matplotlib.pyplot as plt
gender_count = df["gender"].value_counts()

plt.figure(figsize=(4,4))
plt.pie(gender_count.values,
        labels=gender_count.index,
        autopct="%1.1f%%",
        startangle=90)

plt.title("Gender Distribution")
plt.show()
```

```python
# Histplot – Age Distribution
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(6,4))
sns.histplot(df["age"],bins=10)

plt.title("Age Distribution")
plt.xlabel("Age")
plt.ylabel("Number of Customers")

plt.show()
```

```python
# Scatter Plot – Age vs Purchase Amount on basis of gender
import matplotlib.pyplot as plt
plt.figure(figsize=(8,5))
sns.scatterplot(x="age",y="purchase_amount",data=df,hue="gender")

plt.title("Age vs Purchase Amount")
plt.xlabel("Age")
plt.ylabel("Purchase Amount")

plt.show()
```

```python
# Count Plot – Number of Customers by Gender
plt.figure(figsize=(6,4))
sns.countplot(data=df, x="gender")

plt.title("Number of Customers by Gender")
plt.show()
```

```python
#Line Plot – Average Purchase Amount by Age
avg_purchase = df.groupby("age")["purchase_amount"].mean().reset_index()

plt.figure(figsize=(8,5))
sns.lineplot(data=avg_purchase,
             x="age",
             y="purchase_amount",
             marker="o",c="r")

plt.title("Average Purchase Amount by Age")
plt.show()
```

```python
#Violin Plot – Purchase Amount by Gender
plt.figure(figsize=(6,5))
sns.violinplot(data=df,
               x="gender",
               y="purchase_amount")

plt.title("Purchase Amount by Gender")
plt.show()
```

```python
# Heatmap – Correlation Matrix
plt.figure(figsize=(6,5))

corr = df.select_dtypes(include="number").corr()

sns.heatmap(corr,
            annot=True,
            cmap="coolwarm",
            fmt=".2f")

plt.title("Correlation Heatmap")
plt.show()
```

```python
# Pair Plot
sns.pairplot(df.select_dtypes(include="number"))
plt.show()
```
